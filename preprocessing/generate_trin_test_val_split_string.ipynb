```python
import random

def split_cases(items, train_fix_ratio=0.8, train_val_ratio=0.1, test_ratio=0.1, seed=42):
    assert abs(train_fix_ratio + train_val_ratio + test_ratio - 1.0) < 1e-6

    items = list(items)

    random.seed(seed)
    random.shuffle(items)

    n = len(items)

    train_fix_end = int(n * train_fix_ratio)
    train_val_end = train_fix_end + int(n * train_val_ratio)

    split = {
        "train_fix": items[:train_fix_end],
        "train_val": items[train_fix_end:train_val_end],
        "test": items[train_val_end:]
    }

    return split
```

```python
def print_split(split):
    for key, values in split.items():
        print(f"'{key}': {values},")
```

```python
print_split(split_cases(patients, train_fix_ratio=0.8, train_val_ratio=0.1, test_ratio=0.1, seed=42))
```
