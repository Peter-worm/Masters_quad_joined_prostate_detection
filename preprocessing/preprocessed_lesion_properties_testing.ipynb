```python
global_max = {}
global_min = {}

for patient in patients:
    patient_data = file_manager.load_file_pickle(last_step, patient)

    for key, values in patient_data.items():
        current_max = np.max(values)
        current_min = np.min(values)

        if key not in global_max:
            global_max[key] = current_max
            global_min[key] = current_min
        else:
            global_max[key] = max(global_max[key], current_max)
            global_min[key] = min(global_min[key], current_min)

# Overall max/min across all patients
print("\n=== Overall ===")
for key in global_max:
    print(f"{key}: max={global_max[key]} min={global_min[key]}")
```
