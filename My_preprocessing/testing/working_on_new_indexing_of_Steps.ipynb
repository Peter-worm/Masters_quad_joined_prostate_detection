```python
preprocessed_steps = {}
for i,(step, output) in enumerate(preprocessing_steps_list):
    if i == 0:
        continue
    preprocessed_steps[step] = {'start': f"{i-1}_{preprocessing_steps_list[i-1][1]}", 'end': f"{i}_{output}"}
```
